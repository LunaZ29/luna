```python
import pandas as pd
```

```python
df = pd.read_csv("state_data.csv") 
df.head()
```

```python
print(df.shape)
# 4 columns, 52 rows 
```

```python
display(df.head())
display(df.tail())
# diff columns represent self explanatory their title
# first state is alaska, last state is wyoming 
# there's 52 entries total 
# might have more than 50 because of non-state territories 
```

```python
median_lat = df["latitude"].median()
median_lon = df["longitude"].median()

print("Median Latitude:", median_lat)
print("Median Longitude:", median_lon)
```

```python
median_lat = df["latitude"].median()

df["lat_dist_to_median"] = (df["latitude"] - median_lat).abs()

df.head()
```

```python
median_lon = df["longitude"].median()

df["lon_dist_to_median"] = (df["longitude"] - median_lon).abs()

df.head()
```

```python
print(df.shape)
# now there's 6 columns, 52 rows
```

```python
# Lat 
closest_lat = df.loc[df["lat_dist_to_median"].idxmin()]
farthest_lat = df.loc[df["lat_dist_to_median"].idxmax()]

print("Closest to median latitude:", closest_lat["state"], "by", round(closest_lat["lat_dist_to_median"], 2))
print("Farthest from median latitude:", farthest_lat["state"], "by", round(farthest_lat["lat_dist_to_median"], 2))

# Long
closest_lon = df.loc[df["lon_dist_to_median"].idxmin()]
farthest_lon = df.loc[df["lon_dist_to_median"].idxmax()]

print("Closest to median longitude:", closest_lon["state"], "by", round(closest_lon["lon_dist_to_median"], 2))
print("Farthest from median longitude:", farthest_lon["state"], "by", round(farthest_lon["lon_dist_to_median"], 2))
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
crashes_df = sns.load_dataset("car_crashes")
crashes_df.head()
```

```python
crashes_df = crashes_df.dropna(axis=1)
crashes_df.head()
```

```python
print(df.shape)
# 8 columns, 51 rows 
```

```python
display(df.head())
display(df.tail())
# description of each column is self explanatory by title (ins refers to insurance, abbrev refers to state)
```

```python
# 2x2
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Top-left subplot: Scatter plot of two columns
sns.scatterplot(data=crashes_df, x="speeding", y="total", ax=axes[0, 0])
axes[0, 0].set_title("Speeding vs. Total Crashes")

# Top-right subplot: Scatter plot of two completely different columns
sns.scatterplot(data=crashes_df, x="alcohol", y="ins_premium", ax=axes[0, 1])
axes[0, 1].set_title("Alcohol vs. Insurance Premium")

# Bottom-left subplot: Histogram of any column
sns.histplot(data=crashes_df, x="not_distracted", ax=axes[1, 0])
axes[1, 0].set_title("Distribution of Not Distracted Drivers")

# Bottom-right subplot: Boxplot of a different column
sns.boxplot(data=crashes_df, y="ins_losses", ax=axes[1, 1])
axes[1, 1].set_title("Boxplot of Insurance Losses")

plt.tight_layout()
plt.show()
```

```python
# 2x2
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Top-left subplot: Scatter plot (Blue) with a legend
sns.scatterplot(
    data=crashes_df,
    x="speeding",
    y="total",
    ax=axes[0, 0],
    color="blue",
    label="US States"
)
axes[0, 0].set_title("Speeding vs. Total Fatal Collisions")
axes[0, 0].set_xlabel("Drivers Involved in Speeding Crashes (%)")
axes[0, 0].set_ylabel("Total Fatal Collisions (per billion miles)")
axes[0, 0].legend()

# 2. Top-right subplot: Scatter plot of two different columns (Green)
sns.scatterplot(
    data=crashes_df,
    x="alcohol",
    y="ins_premium",
    ax=axes[0, 1],
    color="seagreen"
)
axes[0, 1].set_title("Alcohol Impairment vs. Car Insurance Premiums")
axes[0, 1].set_xlabel("Alcohol-Impaired Drivers in Fatal Crashes (%)")
axes[0, 1].set_ylabel("Car Insurance Premium ($ USD)")

# 3. Bottom-left subplot: Histogram of any column (Coral/Orange)
sns.histplot(
    data=crashes_df,
    x="not_distracted",
    ax=axes[1, 0],
    color="coral"
)
axes[1, 0].set_title("Distribution of Non-Distracted Drivers in Crashes")
axes[1, 0].set_xlabel("Drivers Not Distracted (%)")
axes[1, 0].set_ylabel("Number of States (count)")

# 4. Bottom-right subplot: Boxplot of a different column (Purple)
sns.boxplot(
    data=crashes_df,
    y="ins_losses",
    ax=axes[1, 1],
    color="mediumpurple"
)
axes[1, 1].set_title("Distribution of Insurance Company Losses")
axes[1, 1].set_xlabel("All U.S. States")
axes[1, 1].set_ylabel("Losses Incurred by Insurance Companies ($ USD)")


plt.tight_layout()
plt.show()
```

```python
# Explanations 

# Top-Left: Speeding vs. Total Fatal Collisions: Compares speeding crash percentage vs. total fatal crashes to test if speeding increases overall crash rates 
# Top-Right: Alcohol Impairment vs. Car Insurance Premiums: Compares alcohol-impaired crash percentage vs. average car insurance premiums to see if impaired driving raises insurance costs
# Bottom-Left: Distribution of Non-Distracted Drivers: Shows the distribution of non-distracted drivers in fatal crashes across states to see typical attentiveness levels
# Bottom-Right: Distribution of Insurance Company Losses: Displays the spread and median of insurance company losses to spot typical payout ranges and outliers
```
