```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8], 
    [11, 13, 17],
    [7, 10, 13]
]) 
def prime_rows(arr):
    mask = []
    for row in arr:
        has_prime = False
        for x in row:
            if x >= 2 and all(x % i != 0 for i in range(2, int(x ** 0.5) + 1)):
                has_prime = True
                break
        mask.append(has_prime)
    return arr[mask]

print(prime_rows(arr))
```

```python
def atebyate(): 
    return np.zeros((8,8))
print(atebyate())

def ones(atebyate): 
    for row in range(8): 
        if row % 2 == 1: 
            for col in range(8): 
                if col % 2 == 0: 
                    atebyate[row][col] = 1
    return atebyate
print(ones(atebyate()))

def otherones(atebyate): 
    for row in range(8): 
        if row % 2 == 0: 
            for col in range(8): 
                if col % 2 == 1: 
                    atebyate[row][col] = 1 
    return atebyate 
print(otherones(atebyate()))

def checkerboard(atebyate): 
    ones(atebyate)
    otherones(atebyate)
    return atebyate 

print(checkerboard(atebyate()))

def reverse_checkerboard(atebyate): 
    return 1 - checkerboard(atebyate) 

print(reverse_checkerboard(atebyate()))
```

```python
world = np.array(['Earth', 'Mars']) 
def expansion(world, n): 
    result = [] 
    for word in world: 
        result.append((' ' * n).join(word)) 
    return np.array(result)  

print(expansion(world, 1)) 
print(expansion(world, 2))
```

```python
np.random.seed(123) 
stars = np.random.randint(500, 2000, (5, 5)) 
print(stars) 

def second_dimmest(stars): 
    result = [] 
    for row in stars:
        sorted_row = np.sort(row)
        result.append(sorted_row[-2])
    return np.array(result)

print(second_dimmest(stars))
```
