```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# sin thing
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

# x array (0 to 2π with 1000 points)
x = np.linspace(0, 2 * np.pi, 1000)

amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])

fig, axes = plt.subplots(1, 3, figsize=(25, 5))

# Loop thing
for A, w in zip(amplitude, frequency):
    # First subplot (index 0) 
    y1 = make_sine_wave(x, A, 1)
    axes[0].plot(x, y1, label=f"A = {A}")
    
    # Second subplot (index 1)
    y2 = make_sine_wave(x, 1, w)
    axes[1].plot(x, y2, label=f"w = {w}")
    
    # Third subplot (index 2) w/ both 
    y3 = make_sine_wave(x, A, w)
    axes[2].plot(x, y3, label=f"A = {A}, w = {w}")

axes[0].set_title("Varying Amplitude (w = 1)")
axes[0].set_xlabel("x angle (rads)")
axes[0].set_ylabel("Displacement: A · sin(x)")
axes[0].legend()

axes[1].set_title("Varying Angular Frequency (A = 1)")
axes[1].set_xlabel("x angle (rads)")
axes[1].set_ylabel("Displacement: sin(wx)")
axes[1].legend()

axes[2].set_title("Varying Both Amplitude and Frequency")
axes[2].set_xlabel("x angle (rads)")
axes[2].set_ylabel("Displacement: A · sin(wx)")
axes[2].legend()

plt.tight_layout()
plt.show()
```
