```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
data = np.loadtxt("star_data.txt", skiprows=1)
```

```python
print(np.array(data).shape)
# 2642 rows 
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
print("Min magnitude:", np.min(magnitude))
print("Max magnitude:", np.max(magnitude))

print("Row index of min magnitude:", np.argmin(magnitude))
print("Row index of max magnitude:", np.argmax(magnitude))

print("Min temperature:", np.min(temperature))
print("Max temperature:", np.max(temperature))

print("Row index of min temperature:", np.argmin(temperature))
print("Row index of max temperature:", np.argmax(temperature))
```

```python
print("Brightest star index:", np.argmin(magnitude))
print("Dimmest star index:", np.argmax(magnitude))

# the star in row 1988 is brightest and star in row 264 is dimmest
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

brightest = np.argmin(magnitude)    # Index 1988
dimmest = np.argmax(magnitude)      # Index 264
coldest = np.argmin(temperature)    # Index 1923
hottest = np.argmax(temperature)    # Index 2613

plt.figure(figsize=(8, 6))

plt.scatter(log_temperature, log_magnitude, c=log_temperature, cmap="coolwarm_r", alpha=0.6, label="Stars")

plt.scatter(log_temperature[brightest], log_magnitude[brightest], 
            color="black", marker="*", s=300, label="Brightest Star")

plt.scatter(log_temperature[dimmest], log_magnitude[dimmest], 
            color="purple", marker="*", s=300, label="Dimmest Star")

plt.scatter(log_temperature[hottest], log_magnitude[hottest], 
            color="green", marker="*", s=300, label="Hottest Star")

plt.scatter(log_temperature[coldest], log_magnitude[coldest], 
            color="orange", marker="*", s=300, label="Coldest Star")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")

plt.colorbar(label="Log Temperature")

plt.show()
```
